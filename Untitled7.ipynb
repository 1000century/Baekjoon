```python
import pandas as pd
import torch
from torch.utils.data import Dataset
from datasets import load_dataset
from sklearn.model_selection import train_test_split
from transformers import BertTokenizer

# 1. 데이터셋 로드
news_dataset = load_dataset('ag_news')

# 2. train 데이터를 DataFrame으로 변환
df_train = pd.DataFrame(news_dataset['train'])
test_data = news_dataset['test']

# 3. train 데이터를 train/val/test로 분할
train_texts, temp_texts, train_labels, temp_labels = train_test_split(
    df_train['text'],
    df_train['label'],
    test_size=0.3,
    random_state=42
)

val_texts, test_texts, val_labels, test_labels = train_test_split(
    temp_texts,
    temp_labels,
    test_size=0.5,
    random_state=42
)

# 4. 토크나이저 로드
tokenizer = BertTokenizer.from_pretrained("bert-base-uncased")
# 학습 데이터 토크나이징
train_encodings = tokenizer(
   train_texts.tolist(),
   truncation=True,
   padding='max_length',
   max_length=128,
   return_tensors='pt'
)

# 검증 데이터 토크나이징
val_encodings = tokenizer(
   val_texts.tolist(),
   truncation=True,
   padding='max_length',
   max_length=128,
   return_tensors='pt'
)

# 테스트 데이터 토크나이징
test_encodings = tokenizer(
   test_texts.tolist(),
   truncation=True,
   padding='max_length',
   max_length=128,
   return_tensors='pt'
)
```

```python
# 첫 번째 문장을 디코딩하여 [CLS]와 [SEP] 토큰이 포함되어 있는지 확인
print("First encoded input_ids:", train_encodings['input_ids'][0])

# [CLS]와 [SEP]의 인덱스 값 확인하기
cls_token_id = tokenizer.cls_token_id
sep_token_id = tokenizer.sep_token_id
print(f"[CLS] 토큰 : {cls_token_id}, [SEP] 토큰 : {sep_token_id}")

# 첫 번째 문장의 디코딩된 형태 확인
tokens = tokenizer.convert_ids_to_tokens(train_encodings['input_ids'][0])
decoded_text = " ".join(tokens)
print("디코딩된 문장:", decoded_text)
```

```python
# 5. Dataset 클래스 정의
class NewsDataset(Dataset):
    def __init__(self, texts, labels, tokenizer, max_length=128):
        self.texts = texts
        self.labels = labels
        self.tokenizer = tokenizer
        self.max_length = max_length

    def __len__(self):
        return len(self.texts)

    def __getitem__(self, idx):
        text = str(self.texts.iloc[idx] if hasattr(self.texts, 'iloc') else self.texts[idx])
        label = self.labels.iloc[idx] if hasattr(self.labels, 'iloc') else self.labels[idx]

        encoding = self.tokenizer(
            text,
            truncation=True,
            padding='max_length',
            max_length=self.max_length,
            return_tensors='pt'
        )

        return {
            'input_ids': encoding['input_ids'].flatten(),
            'attention_mask': encoding['attention_mask'].flatten(),
            'label': torch.tensor(label, dtype=torch.long)
        }

# 6. Dataset 인스턴스 생성
train_dataset = NewsDataset(train_texts, train_labels, tokenizer)
val_dataset = NewsDataset(val_texts, val_labels, tokenizer)
test_dataset = NewsDataset(test_texts, test_labels, tokenizer)

# 데이터셋 크기 확인
print(f"Training set size: {len(train_dataset)}")
print(f"Validation set size: {len(val_dataset)}")
print(f"Test set size: {len(test_dataset)}")
```

BERT 구현


```python
import torch.nn as nn
from transformers import BertModel
class BERTClassifier(nn.Module):
    def __init__(self, dropout=0.5):
        super().__init__()
        self.bert = BertModel.from_pretrained("bert-base-uncased")
        self.dropout = nn.Dropout(dropout)
        self.classifier = nn.Linear(self.bert.config.hidden_size, 4)

    def forward(self, input_ids, attention_mask):
        outputs = self.bert(input_ids=input_ids, attention_mask=attention_mask)
        pooled_output = outputs.pooler_output
        dropout_output = self.dropout(pooled_output)
        return self.classifier(dropout_output)
criterion = nn.CrossEntropyLoss()
model = BERTClassifier()

```

```python
from torch.utils.data import DataLoader
from torch.optim import AdamW
from tqdm import tqdm
import numpy as np

def train_model(model, train_dataset, val_dataset, epochs=4, batch_size=128, learning_rate=2e-5):
    # DataLoader 설정
    train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
    val_loader = DataLoader(val_dataset, batch_size=batch_size)

    # 옵티마이저와 loss function 설정
    optimizer = AdamW(model.parameters(), lr=learning_rate)
    criterion = nn.CrossEntropyLoss()

    # GPU 사용 가능시 GPU 사용
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    model = model.to(device)

    best_val_loss = float('inf')

    for epoch in range(epochs):
        # 학습 모드
        model.train()
        total_train_loss = 0
        train_steps = 0

        # 프로그레스바로 학습 과정 표시
        train_pbar = tqdm(train_loader, desc=f'Epoch {epoch+1}/{epochs} [Train]')

        for batch in train_pbar:
            # 배치 데이터를 GPU로 이동
            input_ids = batch['input_ids'].to(device)
            attention_mask = batch['attention_mask'].to(device)
            labels = batch['label'].to(device)

            # Forward pass
            optimizer.zero_grad()
            outputs = model(input_ids, attention_mask)
            loss = criterion(outputs, labels)

            # Backward pass
            loss.backward()
            optimizer.step()

            total_train_loss += loss.item()
            train_steps += 1

            # 프로그레스바 업데이트
            train_pbar.set_postfix({'loss': f'{loss.item():.4f}'})

        avg_train_loss = total_train_loss / train_steps

        # 검증 모드
        model.eval()
        total_val_loss = 0
        val_steps = 0
        correct = 0
        total = 0

        # 검증 시에는 그래디언트 계산 안 함
        with torch.no_grad():
            val_pbar = tqdm(val_loader, desc=f'Epoch {epoch+1}/{epochs} [Val]')

            for batch in val_pbar:
                input_ids = batch['input_ids'].to(device)
                attention_mask = batch['attention_mask'].to(device)
                labels = batch['label'].to(device)

                outputs = model(input_ids, attention_mask)
                loss = criterion(outputs, labels)

                total_val_loss += loss.item()
                val_steps += 1

                # 정확도 계산
                _, predicted = torch.max(outputs.data, 1)
                total += labels.size(0)
                correct += (predicted == labels).sum().item()

                # 프로그레스바 업데이트
                val_pbar.set_postfix({'loss': f'{loss.item():.4f}'})

        avg_val_loss = total_val_loss / val_steps
        accuracy = 100 * correct / total

        print(f'\nEpoch {epoch+1}/{epochs}:')
        print(f'Average training loss: {avg_train_loss:.4f}')
        print(f'Average validation loss: {avg_val_loss:.4f}')
        print(f'Validation Accuracy: {accuracy:.2f}%')

        # 최고 성능 모델 저장
        if avg_val_loss < best_val_loss:
            best_val_loss = avg_val_loss
            torch.save(model.state_dict(), 'best_model.pt')
            print("Best model saved!")

        print("-" * 60)

    return model

# 모델 학습 실행
model = BERTClassifier()
trained_model = train_model(
    model=model,
    train_dataset=train_dataset,
    val_dataset=val_dataset,
    epochs=4,
    batch_size=128,
    learning_rate=2e-5
)
```

```python
import torch
from transformers import BertTokenizer

def predict_text(model, text, max_length=128):
    # 모델을 평가 모드로 설정
    model.eval()

    # GPU 사용 가능시 GPU 사용
    device = torch.device("cpu")

    model = model.cpu()

    # 토크나이저 초기화 (BERT 모델에 맞는 토크나이저 사용)
    tokenizer = BertTokenizer.from_pretrained('klue/bert-base')

    # 텍스트 전처리
    encoding = tokenizer.encode_plus(
        text,
        add_special_tokens=True,
        max_length=max_length,
        padding='max_length',
        truncation=True,
        return_attention_mask=True,
        return_tensors='pt'
    )

    # 입력 데이터를 GPU로 이동
    print(encoding['input_ids'])
    input_ids = encoding['input_ids'].to(device)
    attention_mask = encoding['attention_mask'].to(device)

    # 추론
    with torch.no_grad():
        outputs = model(input_ids, attention_mask)

        # softmax를 통해 확률값으로 변환
        probabilities = torch.nn.functional.softmax(outputs, dim=1)

        # 가장 높은 확률을 가진 클래스 선택
        predicted_class = torch.argmax(probabilities, dim=1)
        confidence = probabilities[0][predicted_class].item()

    return {
        'predicted_class': predicted_class.item(),
        'confidence': confidence * 100  # 백분율로 변환
    }

# 사용 예시
def test_model(model, text):
    result = predict_text(model, text)
    print(f"입력 텍스트: {text}")
    print(f"예측 클래스: {result['predicted_class']}")
    print(f"확신도: {result['confidence']:.2f}%")
    print("-" * 50)
  # 테스트할 텍스트 입력
test_text = "Putin launched Nuclear weapons"
test_model(model, test_text)

test_text = "Messi scored in soccer match again"
test_model(model, test_text)


test_text = "The United Nations calls for immediate ceasefire in the Middle East conflict."
test_model(model, test_text)


test_text ="Lakers win NBA championship after thrilling overtime victory."
test_model(model, test_text)


test_text = "Tesla stock soars 15% after revealing record quarterly profits."
test_model(model, test_text)


test_text =   "Scientists discover new species of deep-sea creatures using advanced underwater robots."
test_model(model, test_text)

```
